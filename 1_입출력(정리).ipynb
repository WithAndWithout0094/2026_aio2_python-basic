# 변수의 유효 범위

```python
# 지역 변수 예시
def local_example():
    local_var = "이것은 지역 변수입니다"
    print(local_var)  # 함수 내에서 접근 가능

# 다른 def에서 현재 def 내 변수 등 요소에 접근 못함
# local_var를 함수 밖에서 접근하면 에러 발생

# 전역 변수 예시
global_var = "이것은 전역 변수입니다"

def global_example():
    print(global_var)  # 전역 변수는 함수 내에서도 접근 가능

# 정적 변수 예시
class Example:
    static_var = "이것은 정적 변수입니다"  # 클래스 변수(정적 변수)

    def print_static():
        print(Example.static_var)  # 클래스를 통해 접근 가능
```

```python
# global 키워드 사용 예시
counter = 0  # 전역 변수

def increment_counter():
    global counter  # global 키워드로 전역 변수 사용 선언
    counter += 1
    print(f"카운터: {counter}")

increment_counter()  # 출력: 카운터: 1
increment_counter()  # 출력: 카운터: 2
```

# 리스트의 특징

```python
#직접 반복 처리
data = [1, -2, 3, -4, 5, -6]
for x in data:
    if x < 0:
        data.remove(x)
print(data)


#누적후 처리
data = [1, -2, 3, -4, 5, -6]
filtered = []
for x in data:
    if x >= 0:
        filtered.append(x)
print(filtered)

# 파이썬 스러운 코드 : 컴프리헨션
data = [1, -2, 3, -4, 5, -6]
filtered = [x for x in data if x >= 0]
print(filtered)
```

# 조건문을 활용한 접근 권한 제어

```python
def check_access(user_role):
    if user_role == "admin":
        return "관리자 페이지에 접근 가능합니다."
    elif user_role == "user":
        return "일반 사용자 페이지에 접근 가능합니다."
    else:
        return "접근 권한이 없습니다."

# 테스트
print(check_access("admin"))   # 관리자 페이지에 접근 가능합니다.
print(check_access("user"))    # 일반 사용자 페이지에 접근 가능합니다.
print(check_access("guest"))   # 접근 권한이 없습니다.
```

# break/continue/pass 문

```python
numbers = [1, 3, 5, 7, 9, 11, 13, 15]
search_number = 9

for num in numbers:
    if num == search_number:
        print(f"{search_number}를 찾았습니다!")
        break
    print(f"현재 검사 중인 숫자: {num}")
```
