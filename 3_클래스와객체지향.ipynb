```python
import random

class Account:

    bank_name = 'SC은행'

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
        self.account_no = Account.make_account_no()

    @staticmethod
    def make_account_no():
        num1 = random.randint(0, 999)
        num2 = random.randint(0,99)
        num3 = random.randint(0,999999)

        return f"{num1:03d}-{num2:02d}-{num3:06d}"

    def print_info(self):
        return f'은행이름 :{Account.bank_name} \n 계좌번호: {self.account_no}'

    def 입금(self, amount):
        self.balance += amount

    def 출금(self,amount):
        self.balance -= amount

    def 잔고조회(self):
        return f'{self.name}님의 잔고는 {self.balance}원 입니다.'
```

```python
ac1 = Account('홍길동', 2000)
print(ac1.print_info())
print(ac1.잔고조회())
print()

ac1.입금(3000)  
print(ac1.print_info())
print(ac1.잔고조회())
print()

ac1.출금
print()

ac2 = Account('바바라',700000000000)
print(ac2.print_info())
print(ac2.잔고조회())
print()
```

```python
class Animal:
    def make_sound(self):
        pass


class Dog(Animal):
    def make_sound(self):
        return "멍멍!"


class Cat(Animal):
    def make_sound(self):
        return "야옹!"
```

# 다양성

```python
animal = [Dog(), Cat(),Dog(), Cat()]


for ani in animal:
    print(ani.make_sound())
```

# 예외 클래스 정의

```python
class AgeException(Exception):
    def __init__(self, msg):
        super().__init__(msg)
        self.message = msg
```

```python
def input_age():
    value = input("나이를 입력하세요: ").strip()
    # 나이를 입력 받아야함


    if value =="":
        raise AgeException("나이 입력하세요.")
        age = int(value)
        
    try:
        age = int(value)
    except ValueError:
        raise AgeException("나이는 정수로 입력해야 합니다.")
    #입력값 체크
    if age <=0: 
        raise AgeException("나이는 0이거나 음수가 될 수 없어")
        # 나이는 양수
        #
    elif age >150:
        raise AgeException("무적이냐? 무적이야? 오래 사네?")
        #상한 체크
        return age

try: 
    age = input_age()
    print(f'입력한 나이는 {age}살 입니다.')

except AgeException as error:
    print("입력 오류:", error)


        #위와 같은 구조
```
